```python
import os
import glob
import sys

def verificar_archivos_usuario():
    """Verifica que los archivos específicos del usuario estén presentes"""
    print("🔍 VERIFICANDO ARCHIVOS DEL PROYECTO DEL USUARIO")
    print("=" * 70)
    
    # Archivos específicos del usuario
    archivos_usuario = {
        'config_database.py': 'Configuración de base de datos',
        'ml_model_production.py': 'Modelo de producción',
        'run_ml_system_real_db.py': 'Sistema principal',
        'extract_real_data_with_synthetic_feedback.py': 'Extracción de datos'
    }
    
    # Buscar en diferentes ubicaciones
    ubicaciones = ['/content', '/content/ml-recomendator', os.getcwd()]
    
    archivos_encontrados = {}
    archivos_faltantes = []
    
    print("📋 ARCHIVOS DEL USUARIO:")
    for archivo, descripcion in archivos_usuario.items():
        encontrado = False
        for ubicacion in ubicaciones:
            ruta_completa = os.path.join(ubicacion, archivo)
            if os.path.exists(ruta_completa):
                print(f"   ✅ {archivo} - {descripcion}")
                print(f"      📍 Ubicación: {ruta_completa}")
                archivos_encontrados[archivo] = ruta_completa
                encontrado = True
                break
        
        if not encontrado:
            print(f"   ❌ {archivo} - {descripcion}")
            archivos_faltantes.append(archivo)
    
    # Buscar archivos CSV
    print("\n📋 ARCHIVOS DE DATOS (CSV):")
    archivos_csv = []
    for ubicacion in ubicaciones:
        if os.path.exists(ubicacion):
            csvs = glob.glob(os.path.join(ubicacion, '*.csv'))
            archivos_csv.extend(csvs)
    
    if archivos_csv:
        print(f"   📊 Encontrados {len(archivos_csv)} archivos CSV:")
        for csv in archivos_csv[:5]:  # Mostrar solo los primeros 5
            print(f"      • {os.path.basename(csv)}")
        if len(archivos_csv) > 5:
            print(f"      ... y {len(archivos_csv) - 5} más")
    else:
        print("   📊 No se encontraron archivos CSV (opcional)")
    
    print("\n" + "=" * 70)
    
    if archivos_faltantes:
        print("❌ FALTAN ARCHIVOS NECESARIOS:")
        for archivo in archivos_faltantes:
            print(f"   • {archivo}")
        print("\n💡 SOLUCIÓN: Sube todos los archivos del usuario antes de continuar.")
        return False, archivos_encontrados, archivos_csv
    else:
        print("✅ TODOS LOS ARCHIVOS DEL USUARIO ESTÁN PRESENTES")
        return True, archivos_encontrados, archivos_csv

# Verificar archivos
archivos_ok, archivos_encontrados, archivos_csv = verificar_archivos_usuario()

if not archivos_ok:
    print("\n⚠️ Por favor, sube todos los archivos necesarios antes de continuar.")
else:
    print("\n🎉 ¡Perfecto! Todos los archivos están listos.")
    print("📋 Puedes continuar con la siguiente celda.")
```

## 🚀 Ejecutar Sistema ML

Esta celda ejecutará todo el pipeline del sistema ML:
- 🌐 Conexión a base de datos de **producción** (DigitalOcean)
- 📊 Extracción de datos reales
- 🤖 Entrenamiento del modelo ML
- 📈 Generación de reportes y métricas
- 💾 Guardado de modelos y resultados

**⏱️ Tiempo estimado:** 5-15 minutos dependiendo de la cantidad de datos.

```python
import sys
import os
import traceback
from datetime import datetime

def detectar_archivos_usuario():
    """
    Detecta automáticamente los archivos específicos del usuario subidos a Colab
    """
    import os
    import glob
    
    print("🔍 Detectando archivos del usuario...")
    
    # Archivos específicos del usuario
    archivos_usuario = {
        'config_database.py': None,
        'ml_model_production.py': None,
        'run_ml_system_real_db.py': None,
        'extract_real_data_with_synthetic_feedback.py': None
    }
    
    # Buscar en múltiples ubicaciones
    ubicaciones = ['/content', '/content/sample_data', '/content/drive/MyDrive']
    
    for ubicacion in ubicaciones:
        if os.path.exists(ubicacion):
            for archivo in archivos_usuario.keys():
                patron = os.path.join(ubicacion, '**', archivo)
                encontrados = glob.glob(patron, recursive=True)
                if encontrados and archivos_usuario[archivo] is None:
                    archivos_usuario[archivo] = encontrados[0]
                    print(f"✅ Encontrado: {archivo} en {encontrados[0]}")
    
    # Detectar archivos CSV
    archivos_csv = []
    for ubicacion in ubicaciones:
        if os.path.exists(ubicacion):
            patron_csv = os.path.join(ubicacion, '**', '*.csv')
            csvs_encontrados = glob.glob(patron_csv, recursive=True)
            archivos_csv.extend(csvs_encontrados)
    
    if archivos_csv:
        print(f"📊 Encontrados {len(archivos_csv)} archivos CSV:")
        for csv in archivos_csv[:5]:  # Mostrar solo los primeros 5
            print(f"   📄 {csv}")
        if len(archivos_csv) > 5:
            print(f"   ... y {len(archivos_csv) - 5} más")
    
    return archivos_usuario, archivos_csv

# Detectar archivos del usuario
archivos_usuario, archivos_csv = detectar_archivos_usuario()

# Verificar que todos los archivos necesarios estén presentes
archivos_faltantes = [archivo for archivo, ruta in archivos_usuario.items() if ruta is None]

if archivos_faltantes:
    print("\n❌ FALTAN ARCHIVOS NECESARIOS:")
    for archivo in archivos_faltantes:
        print(f"   • {archivo}")
    print("\n💡 Por favor, sube todos los archivos necesarios antes de continuar.")
else:
    print("\n✅ Todos los archivos del usuario están presentes")
    print("\n" + "=" * 70)
    print("🚀 INICIANDO ENTRENAMIENTO DEL MODELO")
    print("=" * 70)
    print(f"⏰ Hora de inicio: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    
    try:
        # Agregar el directorio actual al path de Python
        if '/content' not in sys.path:
            sys.path.insert(0, '/content')
        
        print("\n📦 Importando módulos del usuario...")
        
        # Importar el sistema principal
        print("   🔄 Importando run_ml_system_real_db...")
        import run_ml_system_real_db
        
        print("   🔄 Importando ml_model_production...")
        import ml_model_production
        
        print("   🔄 Importando extract_real_data_with_synthetic_feedback...")
        import extract_real_data_with_synthetic_feedback
        
        print("   🔄 Importando config_database...")
        import config_database
        
        print("✅ Módulos del usuario importados correctamente")
        print("📋 Archivos del proyecto detectados:")
        for archivo, ruta in archivos_usuario.items():
            print(f"   • {archivo}")
        
        if archivos_csv:
            print(f"📊 Archivos CSV disponibles: {len(archivos_csv)}")
        
        print("\n" + "=" * 70)
        print("🔥 EJECUTANDO SISTEMA DE ML CON BASE DE DATOS REAL")
        print("=" * 70)
        print("⏳ Este proceso puede tomar varios minutos...")
        print("🌐 Conectando a base de datos de PRODUCCIÓN (DigitalOcean)...")
        print("-" * 70)
        
        # Ejecutar el sistema principal del usuario con configuración de PRODUCCIÓN
        print("🔄 Ejecutando sistema ML con configuración de PRODUCCIÓN...")
        
        # Crear pipeline con configuración de producción
        pipeline = run_ml_system_real_db.MLSystemPipeline('production')
        resultado = pipeline.run_complete_pipeline()
        
        print("\n" + "=" * 70)
        print("🎉 ¡SISTEMA EJECUTADO EXITOSAMENTE!")
        print("=" * 70)
        print(f"⏰ Hora de finalización: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
        
        # Mostrar información del resultado
        if resultado:
            print(f"📊 Resultado: ✅ Pipeline ejecutado exitosamente")
        else:
            print(f"📊 Resultado: ⚠️ Pipeline completado con algunos errores")
        
    except ImportError as e:
        print(f"\n❌ ERROR DE IMPORTACIÓN: {e}")
        print("💡 SOLUCIÓN: Verifica que hayas subido todos los archivos del usuario")
        print(f"📁 Archivos detectados: {list(archivos_usuario.keys())}")
        print("\n🔍 DIAGNÓSTICO:")
        print(f"   📂 Directorio actual: {os.getcwd()}")
        print(f"   🐍 Python path: {sys.path[:3]}...")
        
    except Exception as e:
        print(f"\n❌ ERROR DURANTE LA EJECUCIÓN: {e}")
        print("\n🔍 INFORMACIÓN DE DEBUG:")
        print(f"   📁 Archivos detectados: {list(archivos_usuario.keys())}")
        print(f"   📂 Directorio actual: {os.getcwd()}")
        
        try:
            archivos_dir = os.listdir('.')
            print(f"   📋 Archivos en directorio: {archivos_dir[:10]}{'...' if len(archivos_dir) > 10 else ''}")
        except:
            print("   📋 No se pudo listar el directorio")
        
        print("\n📋 TRACEBACK COMPLETO:")
        print("-" * 50)
        traceback.print_exc()
        print("-" * 50)
    
    # Mostrar resumen de archivos generados
    print("\n" + "=" * 70)
    print("📁 RESUMEN DE ARCHIVOS GENERADOS")
    print("=" * 70)
    
    archivos_esperados = {
        'production_model_report.txt': 'Reporte del modelo',
        'trained_model.pkl': 'Modelo entrenado',
        'model_metadata.json': 'Metadatos del modelo',
        'plots': 'Directorio de gráficos',
        'models': 'Directorio de modelos',
        'data': 'Directorio de datos',
        'sample_data': 'Datos de muestra'
    }

    archivos_encontrados_gen = 0
    for archivo, descripcion in archivos_esperados.items():
        if os.path.exists(archivo):
            archivos_encontrados_gen += 1
            if os.path.isdir(archivo):
                try:
                    contenido = os.listdir(archivo)
                    if archivo == 'plots':
                        plots = [f for f in contenido if f.endswith(('.png', '.jpg', '.jpeg'))]
                        print(f"   ✅ {archivo}/ - {descripcion} ({len(plots)} gráficos)")
                        for plot in plots[:3]:
                            print(f"      📊 {plot}")
                        if len(plots) > 3:
                            print(f"      ... y {len(plots) - 3} más")
                    else:
                        print(f"   ✅ {archivo}/ - {descripcion} ({len(contenido)} archivos)")
                except:
                    print(f"   ✅ {archivo}/ - {descripcion}")
            else:
                try:
                    size = os.path.getsize(archivo)
                    size_str = f"{size:,} bytes" if size < 1024 else f"{size/1024:.1f} KB"
                    print(f"   ✅ {archivo} - {descripcion} ({size_str})")
                except:
                    print(f"   ✅ {archivo} - {descripcion}")
        else:
            print(f"   ❌ {archivo} - {descripcion} (no generado)")
    
    print(f"\n📊 RESUMEN: {archivos_encontrados_gen}/{len(archivos_esperados)} archivos generados")
    
    if archivos_encontrados_gen >= len(archivos_esperados) * 0.5:
        print("🎉 ¡Ejecución exitosa! Los archivos principales se generaron correctamente")
    else:
        print("⚠️ Algunos archivos no se generaron. Revisa los logs para más detalles.")
    
    print("📋 Puedes ver los resultados en las siguientes celdas")
```

## 📊 Ver Resultados

Visualiza los resultados del entrenamiento del modelo.

```python
import os
import json
from IPython.display import Image, display
import matplotlib.pyplot as plt

print("📊 VISUALIZANDO RESULTADOS DEL MODELO")
print("=" * 50)

# 1. Mostrar reporte del modelo si existe
reportes_posibles = [
    'production_model_report.txt',
    'reports/system_execution_report.txt',
    'models/production_menu_recommender_metadata.json'
]

for reporte in reportes_posibles:
    if os.path.exists(reporte):
        print(f"\n📋 CONTENIDO DE {reporte}:")
        print("-" * 40)
        try:
            with open(reporte, 'r', encoding='utf-8') as f:
                contenido = f.read()
                if reporte.endswith('.json'):
                    # Formatear JSON
                    data = json.loads(contenido)
                    print(json.dumps(data, indent=2, ensure_ascii=False))
                else:
                    # Mostrar solo las primeras 50 líneas
                    lineas = contenido.split('\n')
                    for i, linea in enumerate(lineas[:50]):
                        print(linea)
                    if len(lineas) > 50:
                        print(f"\n... y {len(lineas) - 50} líneas más")
        except Exception as e:
            print(f"❌ Error leyendo {reporte}: {e}")
        print("-" * 40)

# 2. Mostrar gráficos si existen
if os.path.exists('plots'):
    print("\n📈 GRÁFICOS GENERADOS:")
    print("-" * 30)
    
    try:
        archivos_plots = [f for f in os.listdir('plots') if f.endswith(('.png', '.jpg', '.jpeg'))]
        
        if archivos_plots:
            for i, plot_file in enumerate(archivos_plots[:5]):  # Mostrar máximo 5 gráficos
                plot_path = os.path.join('plots', plot_file)
                print(f"\n📊 {plot_file}:")
                try:
                    display(Image(plot_path))
                except:
                    print(f"   ⚠️ No se pudo mostrar {plot_file}")
            
            if len(archivos_plots) > 5:
                print(f"\n... y {len(archivos_plots) - 5} gráficos más en la carpeta 'plots'")
        else:
            print("   📊 No se encontraron gráficos en la carpeta 'plots'")
    except Exception as e:
        print(f"   ❌ Error accediendo a gráficos: {e}")
else:
    print("\n📈 No se encontró carpeta 'plots'")

# 3. Mostrar información de modelos
if os.path.exists('models'):
    print("\n🤖 MODELOS GENERADOS:")
    print("-" * 25)
    
    try:
        archivos_modelos = os.listdir('models')
        for modelo in archivos_modelos:
            modelo_path = os.path.join('models', modelo)
            if os.path.isfile(modelo_path):
                size = os.path.getsize(modelo_path)
                size_str = f"{size:,} bytes" if size < 1024*1024 else f"{size/(1024*1024):.1f} MB"
                print(f"   🤖 {modelo} ({size_str})")
    except Exception as e:
        print(f"   ❌ Error accediendo a modelos: {e}")
else:
    print("\n🤖 No se encontró carpeta 'models'")

print("\n" + "=" * 50)
print("✅ Visualización de resultados completada")
```

## 💾 Descargar Resultados

Descarga los modelos entrenados, reportes y gráficos.

```python
import os
import zipfile
from google.colab import files
import shutil

def crear_zip_resultados():
    """Crea un archivo ZIP con todos los resultados"""
    
    print("📦 CREANDO ARCHIVO ZIP CON RESULTADOS...")
    print("=" * 45)
    
    zip_filename = f"resultados_ml_modelo_{datetime.now().strftime('%Y%m%d_%H%M%S')}.zip"
    
    with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
        archivos_agregados = 0
        
        # Agregar archivos individuales
        archivos_importantes = [
            'production_model_report.txt',
            'trained_model.pkl',
            'model_metadata.json',
            'ml_system.log'
        ]
        
        for archivo in archivos_importantes:
            if os.path.exists(archivo):
                zipf.write(archivo)
                archivos_agregados += 1
                print(f"   ✅ Agregado: {archivo}")
        
        # Agregar directorios completos
        directorios = ['models', 'plots', 'reports', 'data']
        
        for directorio in directorios:
            if os.path.exists(directorio) and os.path.isdir(directorio):
                for root, dirs, files in os.walk(directorio):
                    for file in files:
                        file_path = os.path.join(root, file)
                        arcname = os.path.relpath(file_path, '.')
                        zipf.write(file_path, arcname)
                        archivos_agregados += 1
                print(f"   ✅ Agregado directorio: {directorio}/")
    
    print(f"\n📦 ZIP creado: {zip_filename}")
    print(f"📊 Total de archivos: {archivos_agregados}")
    
    # Mostrar tamaño del archivo
    if os.path.exists(zip_filename):
        size = os.path.getsize(zip_filename)
        size_str = f"{size:,} bytes" if size < 1024*1024 else f"{size/(1024*1024):.1f} MB"
        print(f"📏 Tamaño: {size_str}")
    
    return zip_filename

# Crear y descargar ZIP con todos los resultados
try:
    zip_file = crear_zip_resultados()
    
    print("\n💾 DESCARGANDO ARCHIVO ZIP...")
    files.download(zip_file)
    print("✅ Descarga iniciada")
    
except Exception as e:
    print(f"❌ Error creando ZIP: {e}")

print("\n" + "=" * 50)
print("📋 ARCHIVOS DISPONIBLES PARA DESCARGA INDIVIDUAL:")
print("=" * 50)

# Mostrar archivos disponibles para descarga individual
archivos_descarga = [
    ('production_model_report.txt', 'Reporte del modelo'),
    ('trained_model.pkl', 'Modelo entrenado'),
    ('model_metadata.json', 'Metadatos del modelo'),
    ('ml_system.log', 'Log del sistema')
]

for archivo, descripcion in archivos_descarga:
    if os.path.exists(archivo):
        print(f"✅ {archivo} - {descripcion}")
        print(f"   💾 Para descargar: files.download('{archivo}')")
    else:
        print(f"❌ {archivo} - {descripcion} (no disponible)")

# Mostrar directorios disponibles
directorios_descarga = ['models', 'plots', 'reports', 'data']
print("\n📁 DIRECTORIOS DISPONIBLES:")
for directorio in directorios_descarga:
    if os.path.exists(directorio) and os.path.isdir(directorio):
        try:
            archivos = os.listdir(directorio)
            print(f"✅ {directorio}/ ({len(archivos)} archivos)")
        except:
            print(f"✅ {directorio}/")
    else:
        print(f"❌ {directorio}/ (no disponible)")

print("\n🎉 ¡Descarga completada! Revisa tu carpeta de descargas.")
```
